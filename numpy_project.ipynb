```python

import numpy as np
np.random.seed(42)
subjects = np.array(["Python", "Java", "Maths", "DBMS", "English"])
marks = np.random.randint(35, 100, (50, 5))
students = np.array([f"Student {i}" for i in range(1, 51)])
print("STUDENT MARKS ANALYZER")
print("\nMarks of First 10 Students:")
print(marks[:10])
print("\nArray Information")
print("Shape:", marks.shape)
print("Dimensions:", marks.ndim)
print("Total Marks Entries:", marks.size)
print("Data Type:", marks.dtype)
print("\nMarks in Last Subject:")
print(marks[:, -1])
print("\nMarks in First Three Subjects:")
print(marks[:, :3])
total_marks = np.sum(marks, axis=1)
average_marks = np.mean(marks, axis=1)
print("\nStudent Averages:")
print(np.round(average_marks, 2))
print("\nTopper of the Class:")
topper_index = np.argmax(total_marks)
print("Name:", students[topper_index])
print("Total Marks:", total_marks[topper_index])
print("\nSubject-wise Averages:")
subject_averages = np.mean(marks, axis=0)
for i in range(len(subjects)):
    print(subjects[i], ":", round(subject_averages[i], 2))
print("\nSubject-wise Toppers:")
for i in range(len(subjects)):
    topper = np.argmax(marks[:, i])
    print(subjects[i], ":", students[topper], "-", marks[topper, i])
print("\nStudents with Marks Below 40:")
failures = np.any(marks < 40, axis=1)
print(students[failures])
print("Total Failures:", np.sum(failures))
print("\nStudents Scoring Above 90:")
high_scorers = np.any(marks > 90, axis=1)
print(students[high_scorers])
print("\nGrade Analysis:")
A = average_marks >= 90
B = (average_marks >= 75) & (average_marks < 90)
C = (average_marks >= 60) & (average_marks < 75)
D = (average_marks >= 40) & (average_marks < 60)
F = average_marks < 40
print("Grade A:", np.sum(A))
print("Grade B:", np.sum(B))
print("Grade C:", np.sum(C))
print("Grade D:", np.sum(D))
print("Grade F:", np.sum(F))
passed = np.all(marks >= 40, axis=1)
pass_percentage = np.mean(passed) * 100
print("\nPass Percentage:", round(pass_percentage, 2), "%")
```
